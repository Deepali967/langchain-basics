#### Sequential chains ####

```python
from dotenv import load_dotenv
import os

load_dotenv('.env')
print('OPENAI_API_KEY loaded:', bool(os.getenv('OPENAI_API_KEY')))
```

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_openai import ChatOpenAI
from langchain_classic.chains import LLMChain

llm = ChatOpenAI(model_name="gpt-3.5-turbo", temperature=0.9)

prompt_review = ChatPromptTemplate.from_messages(
    [
        ("human", "You ordered a {food} and your experience was {experience}."),
    ]
)

chain_review = LLMChain(llm=llm, prompt=prompt_review, output_key="review")


prompt_comment = ChatPromptTemplate.from_messages([
    ("human", "Given the review: {review}, write a comment to the customer."),
])


chain_comment = LLMChain(llm=llm, prompt=prompt_comment, output_key="comment")


prompt_summary = ChatPromptTemplate.from_messages([
    ("human", "Summarize the following comment: {comment}"),
])


chain_summary = LLMChain(llm=llm, prompt=prompt_summary, output_key="summary")


prompt_translation = ChatPromptTemplate.from_messages([
    ("human", "Translate the following summary into {language}: {summary}"),
])
chain_translation = LLMChain(llm=llm, prompt=prompt_translation, output_key="translation")

```

```python
from langchain_core.runnables import RunnableSequence

sequence = RunnableSequence(
        chain_review | chain_comment | chain_summary | chain_translation
)


result = sequence.invoke({
    "food": "Pizza",
    "experience": "It was delicious and arrived on time.",
    "language": "Spanish",
})


print(result)
```
