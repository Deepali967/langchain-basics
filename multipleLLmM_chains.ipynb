```python
from dotenv import load_dotenv, find_dotenv

load_dotenv(find_dotenv())
```

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_openai import ChatOpenAI
from langchain_classic import LLMChain

llm = chat = ChatOpenAI(model_name="gpt-3.5-turbo", temperature=0.9)

prompt_template = ChatPromptTemplate.from_messages(
[
    ("human", "tell me a joke about {input}")
]
)

chain = LLMChain(llm=llm, prompt=prompt_template)
chain.run("Parrot")
```

#### Multiple Chain with Prompt Templates #####

```python
multiple_chain_prompt_template = ChatPromptTemplate.from_messages(
    [
        ("human", "tell me a joke about {input} in {language}"),

    ]
)

multiple_chain = LLMChain(llm=llm, prompt=multiple_chain_prompt_template)
multiple_chain.run({"input": "Parrot", "language": "hindi"})
```
