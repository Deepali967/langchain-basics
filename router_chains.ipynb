#### Router Chains ####

Instead of defining multiple chains, we can let LLM decide which follow up chain to be used

```python
from dotenv import load_dotenv, find_dotenv
load_dotenv(find_dotenv())
```

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.runnables import RunnableBranch, RunnableLambda
from langchain_openai import ChatOpenAI
from langchain_classic.chains import LLMChain


llm = ChatOpenAI(model_name="gpt-3.5-turbo", temperature=0.9, max_tokens=1000)

template_1 = ChatPromptTemplate.from_messages(
    [
        ("human", "tell me a joke about {input} in {language}"),
    ]
)

template_2 = ChatPromptTemplate.from_messages(
    [
        ("human", "tell me a story about {input} in {language}"),
    ]
)


template_3 = ChatPromptTemplate.from_messages(
    [
        ("human", "tell me a poem about {input} in {language}"),
    ]
)


```

```python
branch = RunnableBranch(
   (lambda x: x["type"] == "joke", LLMChain(llm=llm, prompt=template_1)),
   (lambda x: x["type"] == "story", LLMChain(llm=llm, prompt=template_2)),
   (lambda x: x["type"] == "poem", LLMChain(llm=llm, prompt=template_3)),
    RunnableLambda(lambda x: "I am sorry, I do not understand.")
)   
```

```python
branch.invoke({"type": "poem", "input": "Parrot", "language": "hindi"})
```
